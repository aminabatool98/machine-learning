```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data=pd.read_csv('/kaggle/input/wine-data/wine.data',header=None,usecols=[1,2,3])
data.columns=['Class label','Alcohol','Malic acid']
```

```python
df=data
```

```python
sns.kdeplot(df['Class label'])
plt.show()
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
# color_dict={1:'red',2:'yellow',3:'green'}
sns.scatterplot(data=df,x='Alcohol',y='Malic acid',hue='Class label',palette='summer')
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.drop('Class label',axis=1),df['Class label'],random_state=0,train_size=0.3)
```

```python

from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler()

scaler.fit(X_train)

X_train_scaler=scaler.transform(X_train)
X_test_scaler=scaler.transform(X_test)
```

```python
# converting array to dataframe
X_train_scaler=pd.DataFrame(X_train_scaler,columns=X_train.columns)
X_test_scaler=pd.DataFrame(X_test_scaler,columns=X_test.columns)
```

```python
X_train_scaler.head()
```

```python
np.round(X_train_scaler.describe(),1)
```

```python
np.round(X_train.describe(),2)
```

```python
fig, (ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))

ax1.scatter(X_train['Alcohol'],X_train['Malic acid'],c=y_train)
ax1.set_title("before scailing")
ax2.scatter(X_train_scaler['Alcohol'],X_train_scaler['Malic acid'],c=y_train)
ax2.set_title("after scaling")
plt.show()
```

```python
ig, (ax1,ax2)=plt.subplots(ncols=2, figsize=(10,5))

ax1.set_title("before scaling")
sns.kdeplot(X_train['Alcohol'],ax=ax1)
sns.kdeplot(X_train['Malic acid'],ax=ax1)


ax2.set_title('after Standard scaling')
sns.kdeplot(X_train_scaler['Alcohol'],ax=ax2)
sns.kdeplot(X_train_scaler['Malic acid'],ax=ax2)
plt.show()
```

```python
fig, (ax1,ax2)=plt.subplots(ncols=2,figsize=(10,5))

ax1.set_title('alcohol before standerd distribution')
sns.kdeplot(X_train['Alcohol'],ax=ax1)
ax1.legend("Alcohol")
ax2.set_title("alcohol after stander distribution")
sns.kdeplot(X_train_scaler['Alcohol'],ax=ax2)
ax1.legend("Alcohol")
```
